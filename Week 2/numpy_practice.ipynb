```python
import numpy as np

scores = np.array([
    [85, 90, 78],
    [70, 65, 80],
    [95, 88, 92],
    [60, 75, 68],
    [88, 82, 91]
])
```

##
First Task

```python
print (f"Shape: {scores.shape}")
print (f"Dimensions: {scores.ndim}")
print (f"Data type: {scores.dtype}")
print (f"Number of students: {scores.shape[0]}")
print (f"Number of subjects: {scores.shape[1]}")
```

```python
SumRows = scores.sum(axis=1)
SumColumns = scores.sum(axis=0)
Min = np.min(scores)
Max = np.max(scores)
NoOfStudents = scores.shape[0]
NoOfSubjects = scores.shape[1]
avg_ScoreForEachStudent = np.divide(SumRows,NoOfSubjects)
avg_ScoreForEachSubject = np.divide(SumColumns,NoOfStudents)
subjects = ["math","Physics ","Programming"]

print (f"Highest score in the whole array: {Max}")
print(f"Lowest score in the whole array: {Min}")
print("#"*45)
for n in range(NoOfStudents):
    print (f"Student #{int(n+1)} has average: {avg_ScoreForEachStudent[n]}")

print("#"*45)

for n in range(NoOfSubjects):
    print (f"#{subjects[n]} has average: {avg_ScoreForEachSubject[n]}")

```
