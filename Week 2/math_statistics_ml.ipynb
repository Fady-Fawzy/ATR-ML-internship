```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
scalar = 5

vector = np.array([1,2,3])

matrix = np.array([
    [1,2,3],
    [4,5,6]
])

print (scalar)
print (vector)
print (matrix)
print (matrix.T)
```

```python
a = np.array([1,2,3])
b = np.array([4,5,6])

manual = 1*4+2*5+3*6
numpy = np.dot(a,b)
numpy2 = a@b
print(manual)
print(numpy)
print(numpy2)
```

```python
a = np.array([1, 2])
b = np.array([4, 6])
distance = np.linalg.norm(a - b)
Cosine = np.dot(a,b)/(np.linalg.norm(a) * np.linalg.norm(b))
print(distance)
print(Cosine)
```

```python
data = np.array([1, 2, 3, 4, 10])

print("Mean:", np.mean(data))
print("Median:", np.median(data))
print("Variance:", np.var(data))
print("Standard Deviation:", np.std(data))
```

```python
x = np.array([1, 2, 3, 4, 5])
y = np.array([10, 20, 30, 40, 50])

print("Covariance:")
print(np.cov(x, y))

print("Correlation:")
print(np.corrcoef(x, y))
```

```python
w = np.linspace(-5, 10, 100)

loss = (w - 3) ** 2

plt.plot(w, loss)
plt.xlabel("Weight")
plt.ylabel("Loss")
plt.title("Loss Function")
plt.show()
```

```python
age = np.array([20, 30, 40])
salary = np.array([5000, 50000, 100000])

age_scaled = (age - age.mean()) / age.std()
salary_scaled = (salary - salary.mean()) / salary.std()

print(age_scaled)
print(salary_scaled)
```
