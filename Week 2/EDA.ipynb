# EDA Observations

1. The dataset initially contains **32,561 rows**.

2. The dataset contains both **integer** and **string** data types.

3. Missing values were found in:
   - `workclass`: **1,836**
   - `type_of_job`: **1,843**
   - `native_country`: **583**

   These missing values were handled during data cleaning.

4. `education` and `education_num` provide essentially the same information, which indicates **feature redundancy**. One of them should be removed before modeling.

5. The dataset contained **24 duplicated rows**, which were removed.

6. The target variable `income` shows **class imbalance**, with the majority of records belonging to the `<=50K` class.

7. `capital_gain` and `capital_loss` are highly concentrated around zero and strongly skewed. Therefore, their box plots are not very informative for understanding their distributions.

8. `fnlwgt` is a **sampling weight** and should be handled separately rather than treated blindly as a normal feature.

9. No obvious **data leakage** was detected in the dataset.
